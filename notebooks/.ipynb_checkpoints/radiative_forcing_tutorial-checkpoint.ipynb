# Stratospheric Aerosol Radiative Forcing — An Interactive Tutorial

This notebook walks through the **direct shortwave radiative forcing** (RF) calculation for stratospheric sulfate aerosol, following the simple-model approach of Chylek & Wong (1995) as used by Pierce et al. (2010) in the context of stratospheric aerosol injection (SAI) for climate engineering.

By the end you will:

1. Understand the Chylek & Wong (1995) RF equation and each of its parameters.
2. See how Mie scattering produces the size-dependent scattering efficiency $Q_\text{sca}(r)$ and asymmetry parameter $g(r)$.
3. Compute the upscatter fraction $\beta(g)$ from Wiscombe & Grams (1976).
4. Derive the H$_2$SO$_4$/H$_2$O solution composition from Tabazadeh et al. (1997).
5. Reproduce Pierce et al. (2010) Figure 1 (scattering cooling efficiency vs particle size).
6. Explore how the result changes when you move sliders for $T$, RH, refractive index, and the Pierce-SI parameters.

## 2. Why radiative forcing?

Stratospheric sulfate aerosol — whether produced naturally by volcanic eruptions (e.g. Pinatubo 1991) or hypothetically injected for **stratospheric aerosol injection (SAI)** geoengineering — cools the Earth by backscattering incoming solar radiation to space before it reaches the surface. The *direct* radiative forcing (ignoring cloud interactions) is the quantity we compute here.

Pierce et al. (2010) used a simple analytical model to assess how efficiently different particle sizes cool the planet per unit of injected sulfur. The simple model is fast enough to combine with a microphysical model of particle growth (coagulation + condensation) and scan a wide parameter space of injection strategies. The full chain is:

$$
\text{SO}_2 \to \text{H}_2\text{SO}_4 \text{ gas} \to \text{nucleation + condensation + coagulation}
\to \text{size distribution} \to \text{RF}.
$$

This notebook focuses on the **last step**: given a particle size distribution and composition, what is the RF?

## 3. The Chylek & Wong (1995) RF equation

For a purely scattering aerosol layer in the thin-aerosol limit:

$$
\boxed{\;\text{RF} \;=\; -\frac{S_0}{4}\, T_a^2\,(1-A)\,(1-R)^2\, 2\beta\, \tau\;}
$$

| Symbol | Meaning | Pierce (2010) value |
|---|---|---|
| $S_0$ | Solar constant (total solar irradiance) [W/m²]. $S_0/4$ is the global-average insolation. | **1370** |
| $T_a$ | Atmospheric transmittance **above** the aerosol layer. Stratospheric aerosol has nothing above to attenuate, so $T_a \approx 1$. Tropospheric aerosol uses $T_a \approx 0.85$. | **1.0** |
| $A$ | Cloud fraction. Clouds block the aerosol's scattering effect; only the clear-sky fraction $(1-A)$ contributes to RF. | **0.6** |
| $R$ | Clear-sky surface albedo. Over bright surfaces upscattered light would have been reflected anyway, reducing net cooling. The $(1-R)^2$ factor accounts for the double pass through the aerosol layer above a non-black surface. | **0.15** |
| $\beta$ | Upscatter fraction. Fraction of scattered light redirected back to space. Computed from the Henyey-Greenstein phase function averaged over solar zenith angles (Wiscombe & Grams 1976). | computed |
| $\tau$ | Scattering optical depth of the aerosol layer. | computed |

The negative sign reflects that scattering aerosol **cools** the planet (negative top-of-atmosphere net flux anomaly).

## 4. Mie scattering

For a spherical particle of radius $r$ and complex refractive index $n$ illuminated by light of wavelength $\lambda$, Mie theory solves Maxwell's equations exactly in terms of the **size parameter** $x = 2\pi r / \lambda$ and returns:

- $Q_\text{sca}(x, n)$ — scattering efficiency (dimensionless, typical range 0–5)
- $Q_\text{ext}(x, n)$ — extinction efficiency (scattering + absorption)
- $g(x, n)$ — asymmetry parameter, the mean cosine of the scattering angle ($g=0$ isotropic, $g=1$ pure forward)

Three regimes are visible in the figure below:

- **Rayleigh** ($x \ll 1$): $Q_\text{sca} \sim x^4$, very weak scattering, nearly isotropic ($g \approx 0$).
- **Mie resonance** ($x \sim 1$–10): oscillations due to interference between scattered waves. Peak around $r \sim \lambda/2$.
- **Geometric optics** ($x \gg 1$): $Q_\text{sca} \to 2$ (a particle extinguishes twice its geometric cross-section due to forward diffraction), and $g$ approaches ~1 (strongly forward-scattering).

```python
wavelength = 500e-9  # 500 nm (Pierce default)
refindex = complex(1.4, 1e-8)  # ~65% H2SO4/H2O

radii_mie = np.logspace(-9, -5, 120)  # 1 nm to 10 um
x_arr = 2.0 * np.pi * radii_mie / wavelength

Qext_list, Qsca_list, g_list = [], [], []
for x in x_arr:
    qe, qs, gs = bhmie_qsca_jax(float(x), refindex)
    Qext_list.append(float(qe))
    Qsca_list.append(float(qs))
    g_list.append(float(gs))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
axes[0].loglog(radii_mie * 1e6, Qsca_list, label=r'$Q_{\mathrm{sca}}$')
axes[0].loglog(radii_mie * 1e6, Qext_list, '--', label=r'$Q_{\mathrm{ext}}$')
axes[0].axhline(2.0, color='gray', ls=':', alpha=0.6, label=r'$Q \to 2$ (geom. optics)')
axes[0].set_xlabel('Particle radius [$\\mu$m]')
axes[0].set_ylabel('Efficiency')
axes[0].set_title(f'Mie efficiency ($\\lambda$={wavelength*1e9:.0f} nm, $n$={refindex.real}+{refindex.imag:.0e}i)')
axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].semilogx(radii_mie * 1e6, g_list, color='tab:orange')
axes[1].set_xlabel('Particle radius [$\\mu$m]')
axes[1].set_ylabel('Asymmetry parameter $g$')
axes[1].set_title('Asymmetry $g$ (0 = isotropic, 1 = pure forward)')
axes[1].grid(True, alpha=0.3)
fig.tight_layout()
```

## 5. Upscatter fraction $\beta$

For the simple RF model we need the **fraction of scattered light directed back to space** for a given asymmetry parameter $g$ and solar zenith angle $\theta_0$. Wiscombe & Grams (1976) derived a two-stream expression (their equation 22) by integrating the Henyey-Greenstein phase function over the upward hemisphere:

$$
\beta(g, \theta_0) \;=\; \frac{1}{2\pi} \int_0^{2\pi}\!\!\int_{\pi/2}^{\pi} p_{\mathrm{HG}}(g,\cos\Theta)\, \sin\theta\,\mathrm{d}\theta\,\mathrm{d}\phi ,
$$

where $\cos\Theta$ is the scattering angle relative to the incident direction.

Key limits (overhead sun, $\theta_0=0$):

- $g=0$ (isotropic) $\Rightarrow \beta = 0.5$
- $g=1$ (pure forward) $\Rightarrow \beta = 0$

Stratospheric sulfate has $g \approx 0.6$–0.8, giving $\beta \approx 0.2$–0.3 at an overhead sun — only a fraction of scattered photons escape to space.

In the full calculation we average $\beta$ over 18 latitude bands × 12 months × 24 hours, weighted by insolation, to get a global annual mean. For intuition the cell below just shows $\beta(g)$ at a single SZA.

```python
g_sweep = np.linspace(0.0, 0.99, 60)
sza_values_deg = [0, 30, 60, 75]

fig, ax = plt.subplots(figsize=(8, 4.5))
for sza_deg in sza_values_deg:
    sza_rad = np.deg2rad(sza_deg)
    beta_vals = np.array([float(upscatter_fraction(g, sza_rad)) for g in g_sweep])
    ax.plot(g_sweep, beta_vals, label=f'SZA = {sza_deg}$^\\circ$')

ax.axhline(0.5, color='gray', ls=':', alpha=0.6, label='isotropic limit')
ax.set_xlabel('Asymmetry parameter $g$')
ax.set_ylabel(r'Upscatter fraction $\beta$')
ax.set_title('Wiscombe & Grams (1976) eqn 22: $\\beta(g, \\theta_0)$')
ax.legend(); ax.grid(True, alpha=0.3)
fig.tight_layout()
```

## 6. Aerosol composition — Tabazadeh et al. (1997)

Stratospheric sulfate aerosol is not pure H$_2$SO$_4$; it is a **binary H$_2$SO$_4$/H$_2$O solution** in equilibrium with the ambient water vapour. The weight-percent of H$_2$SO$_4$ depends sensitively on temperature and relative humidity.

Tabazadeh et al. (1997) Table 1 parameterizes the water vapour pressure over the solution as

$$
\ln P_{\mathrm{H_2O}}(T, w) = a(w) + \frac{b(w)}{T} + \frac{c(w)}{T^2}
$$

for 15 weight percents $w \in \{10, 15, \ldots, 80\}\%$. We invert this relation: given $T$ and RH, we solve for the $w$ whose equilibrium vapour pressure matches the ambient $P_{\mathrm{H_2O}}$.

Pierce (2010) assumed 75 wt% (very dry), which corresponds to RH $< 1\%$ at $T = 220$ K. Typical lower-stratosphere RH of 2–10% gives **50–70 wt%** — somewhat more water, lower density, but higher scattering efficiency per unit sulfur because each particle contains more material.

```python
T_grid = np.linspace(195, 255, 40)
RH_grid = np.linspace(1, 20, 40)
wt_heat = np.zeros((len(RH_grid), len(T_grid)))
rho_heat = np.zeros_like(wt_heat)
for i, rh in enumerate(RH_grid):
    for j, T in enumerate(T_grid):
        w = h2so4_equilibrium_wt(T, rh)
        wt_heat[i, j] = w
        rho_heat[i, j] = h2so4_solution_density(w)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
im0 = axes[0].pcolormesh(T_grid, RH_grid, wt_heat, cmap='viridis', shading='auto')
axes[0].set_xlabel('Temperature [K]')
axes[0].set_ylabel('Relative humidity [%]')
axes[0].set_title('H$_2$SO$_4$ weight percent [%]')
plt.colorbar(im0, ax=axes[0])
axes[0].axhline(5, color='white', ls='--', alpha=0.8)
axes[0].axvline(220, color='white', ls='--', alpha=0.8)
axes[0].plot(220, 5, 'w*', ms=14, label='T=220K, RH=5%')
axes[0].legend(loc='upper right')

im1 = axes[1].pcolormesh(T_grid, RH_grid, rho_heat, cmap='plasma', shading='auto')
axes[1].set_xlabel('Temperature [K]')
axes[1].set_ylabel('Relative humidity [%]')
axes[1].set_title('Solution density [kg/m$^3$]')
plt.colorbar(im1, ax=axes[1])
fig.tight_layout()

w0 = h2so4_equilibrium_wt(220.0, 5.0)
r0 = h2so4_solution_density(w0)
print(f'At T=220 K, RH=5%:  {w0:.1f} wt% H2SO4, rho={r0:.0f} kg/m^3')
```

## 7. Single-wavelength vs full-spectrum Mie

Pierce et al. (2010) evaluated Mie scattering at a single wavelength (500 nm). This captures the peak of the solar spectrum but overestimates the peak scattering efficiency because the Mie resonance is sharp in $r$ and gets smoothed out when integrated over the solar spectrum (Planck at 5778 K, 300–2500 nm).

The figure below compares the two approaches at $T=220$ K, RH=5% (Tabazadeh composition).

```python
T_ref = 220.0
RH_ref = 5.0
wt_ref = h2so4_equilibrium_wt(T_ref, RH_ref)
rho_ref = h2so4_solution_density(wt_ref)

pierce_kw = dict(
    refindex=REFINDEX_SULFATE,
    density=rho_ref,
    solar_constant=SOLAR_CONSTANT_PIERCE,
    Tatm=TATM_STRATOSPHERIC,
    albedo=ALBEDO_SURFACE_CLEARSKY,
    cloud_fraction=CLOUD_FRACTION_DEFAULT,
    n_radii=80, r_min=5e-9, r_max=10e-6,
)

radii_s, rf_single = scattering_efficiency_vs_radius(
    wavelength=500e-9, spectral=False, **pierce_kw)
radii_b, rf_spec = scattering_efficiency_vs_radius(
    spectral=True, n_wavelengths=50, **pierce_kw)

S_FRAC = wt_ref / 100.0 * 32.0 / 98.0  # sulfur mass fraction of wet particle
KG_PER_MTS = 1e9 / S_FRAC               # Mt-S of wet aerosol
cool_single = -rf_single * KG_PER_MTS
cool_spec = -rf_spec * KG_PER_MTS

fig, ax = plt.subplots(figsize=(9, 5))
ax.semilogx(radii_s * 1e6, cool_single, color='tab:red', lw=2,
            label=f'Single $\\lambda=500$ nm  (peak={cool_single.max():.2f})')
ax.semilogx(radii_b * 1e6, cool_spec, color='tab:blue', lw=2,
            label=f'Solar-spectrum weighted (50 bands)  (peak={cool_spec.max():.2f})')
ax.set_xlim(5e-3, 10)
ax.set_xlabel('Particle radius [$\\mu$m]')
ax.set_ylabel('Scattering cooling efficiency\n[W m$^{-2}$ per Mt-S]')
ax.set_title(f'Single-$\\lambda$ vs spectral Mie  (T={T_ref:.0f}K, RH={RH_ref:.0f}%, '
             f'{wt_ref:.0f} wt%, $\\rho$={rho_ref:.0f} kg/m$^3$)')
ax.legend(); ax.grid(True, alpha=0.3)
fig.tight_layout()

peak_shift = radii_b[np.argmax(cool_spec)] / radii_s[np.argmax(cool_single)]
mag_ratio = cool_spec.max() / cool_single.max()
print(f'Peak radius shift (spectral / single): {peak_shift:.2f}x')
print(f'Peak magnitude ratio  (spectral / single): {mag_ratio:.2f}x  ({(1-mag_ratio)*100:.0f}% reduction)')
```

## 8. Reproducing Pierce et al. (2010) Figure 1

The full spectral calculation with self-consistent Tabazadeh composition at several RH values. Each curve uses its own wt%-dependent solution density, and RF is normalized to **per megaton sulfur** so curves differ only because:

- Wetter aerosol (higher RH → lower wt%) means each Mt-S is diluted into more total mass, so more scattering cross-section per Mt-S → higher peak.
- The Mie peak shifts slightly with density since radius at fixed mass changes.

The peak occurs at $r \approx 0.23\,\mu$m with spectral weighting (vs $\approx 0.19\,\mu$m at 500 nm), which is the size regime that coagulation/condensation models of SAI try to target.

```python
T_strat = 220.0
rh_values = [2, 5, 10, 20]
colors = ['tab:red', 'tab:blue', 'tab:green', 'tab:orange']

fig, ax = plt.subplots(figsize=(10, 5.5))
for rh, c in zip(rh_values, colors):
    wt = h2so4_equilibrium_wt(T_strat, rh)
    rho = h2so4_solution_density(wt)
    r_vals, rf = scattering_efficiency_vs_radius(
        refindex=REFINDEX_SULFATE,
        density=rho,
        solar_constant=SOLAR_CONSTANT_PIERCE,
        Tatm=TATM_STRATOSPHERIC,
        albedo=ALBEDO_SURFACE_CLEARSKY,
        cloud_fraction=CLOUD_FRACTION_DEFAULT,
        spectral=True, n_wavelengths=50,
        n_radii=80, r_min=5e-9, r_max=10e-6,
    )
    s_frac = wt / 100.0 * 32.0 / 98.0
    cool = -rf * (1e9 / s_frac)
    peak_val = cool.max(); peak_r = r_vals[np.argmax(cool)] * 1e6
    ax.semilogx(r_vals * 1e6, cool, color=c, lw=2,
                label=f'RH={rh}% ({wt:.0f} wt%, $\\rho$={rho:.0f}, '
                      f'peak={peak_val:.2f} @ {peak_r:.2f} $\\mu$m)')

ax.set_xlim(5e-3, 10)
ax.set_xlabel('Particle radius [$\\mu$m]')
ax.set_ylabel('Scattering cooling efficiency\n[W m$^{-2}$ per Mt-S]')
ax.set_title(f'Pierce et al. (2010) Fig 1 — spectral Mie, Tabazadeh composition (T={T_strat:.0f} K)')
ax.legend(fontsize=9); ax.grid(True, alpha=0.3)
fig.tight_layout()
```

## 9. Interactive playground

Drag the sliders to see how the cooling efficiency curve responds to each input. Each move triggers a fresh Mie + Tabazadeh calculation so updates may take ~1 second at `n_wavelengths=50`. Use `n_wavelengths=1` for a near-instant single-wavelength view.

Physical ranges:

- **Temperature** 185–260 K — lower stratosphere to mid-stratosphere.
- **Relative humidity** 1–20% — typical stratospheric range.
- **Refractive index (real)** 1.3–1.6 — lower = more water, higher = more acid.
- **Solar constant** 1360–1380 W/m² — observational range.
- **$T_a$** 0.85 (troposphere) → 1.0 (top of stratosphere, nothing above).
- **Cloud fraction** 0–0.8 — 0 is clear sky, 0.6 is Pierce default.
- **Albedo** 0.05 (ocean) → 0.7 (ice/snow). 0.15 is Pierce default.

```python
def plot_rf_interactive(T, RH, n_real, n_imag_log, solar_constant,
                        Tatm, cloud_fraction, albedo, n_wavelengths):
    wt = h2so4_equilibrium_wt(T, RH)
    rho = h2so4_solution_density(wt)
    nref = complex(n_real, 10.0 ** n_imag_log)
    spectral = n_wavelengths > 1

    radii_pl, rf = scattering_efficiency_vs_radius(
        wavelength=500e-9, refindex=nref, density=rho,
        solar_constant=solar_constant, Tatm=Tatm,
        albedo=albedo, cloud_fraction=cloud_fraction,
        spectral=spectral, n_wavelengths=int(n_wavelengths),
        n_radii=80, r_min=5e-9, r_max=10e-6,
    )
    s_frac = wt / 100.0 * 32.0 / 98.0
    cool = -rf * (1e9 / s_frac)
    peak_val = cool.max(); peak_r_um = radii_pl[np.argmax(cool)] * 1e6

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.8),
                                   gridspec_kw={'width_ratios': [1, 2]})

    ax1.bar(['wt%', 'rho / 1000'], [wt, rho / 1000.0],
            color=['steelblue', 'coral'])
    ax1.set_ylim(0, 2.0)
    ax1.set_title(f'Composition (T={T:.0f} K, RH={RH:.0f}%)')
    ax1.set_ylabel('wt% / 100   or   $\\rho$ [g/cm$^3$]')
    ax1.text(0, wt + 2, f'{wt:.1f}%', ha='center')
    ax1.text(1, rho / 1000.0 + 0.05, f'{rho:.0f}', ha='center')
    ax1.grid(True, axis='y', alpha=0.3)

    ax2.semilogx(radii_pl * 1e6, cool, lw=2.2, color='tab:blue')
    ax2.axvline(peak_r_um, color='gray', ls='--', alpha=0.6)
    ax2.annotate(f'peak {peak_val:.2f} W/m$^2$/Mt-S\n@ r={peak_r_um:.2f} $\\mu$m',
                 xy=(peak_r_um, peak_val),
                 xytext=(peak_r_um * 3, peak_val * 0.8),
                 arrowprops=dict(arrowstyle='->', alpha=0.6),
                 fontsize=10)
    ax2.set_xlim(5e-3, 10)
    ax2.set_xlabel('Particle radius [$\\mu$m]')
    ax2.set_ylabel('Cooling efficiency [W m$^{-2}$ per Mt-S]')
    title_mode = 'spectral' if spectral else 'single $\\lambda$=500nm'
    ax2.set_title(f'RF per Mt-S  ({title_mode})  '
                  f'n={n_real:.2f}+{10.0**n_imag_log:.1e}i')
    ax2.grid(True, alpha=0.3)
    fig.tight_layout()
    plt.show()


controls = widgets.interact(
    plot_rf_interactive,
    T=widgets.FloatSlider(value=220, min=185, max=260, step=2,
                          description='T [K]', continuous_update=False),
    RH=widgets.FloatSlider(value=5, min=1, max=20, step=1,
                           description='RH [%]', continuous_update=False),
    n_real=widgets.FloatSlider(value=1.4, min=1.3, max=1.6, step=0.01,
                               description='n (real)', continuous_update=False),
    n_imag_log=widgets.FloatSlider(value=-8, min=-9, max=-2, step=0.5,
                                   description='log10(n imag)',
                                   continuous_update=False),
    solar_constant=widgets.FloatSlider(value=1370, min=1360, max=1380, step=1,
                                       description='S0 [W/m^2]',
                                       continuous_update=False),
    Tatm=widgets.FloatSlider(value=1.0, min=0.85, max=1.0, step=0.01,
                             description='T_a', continuous_update=False),
    cloud_fraction=widgets.FloatSlider(value=0.6, min=0.0, max=0.8, step=0.05,
                                       description='Cloud A',
                                       continuous_update=False),
    albedo=widgets.FloatSlider(value=0.15, min=0.05, max=0.7, step=0.01,
                               description='Albedo R',
                               continuous_update=False),
    n_wavelengths=widgets.Dropdown(options=[('1 (single)', 1), ('10', 10), ('30', 30), ('50', 50)],
                                   value=30, description='# $\\lambda$'),
);
```
